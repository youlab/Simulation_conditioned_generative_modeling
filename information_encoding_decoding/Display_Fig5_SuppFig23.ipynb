```python

import torch 
import os 
import cv2 
import matplotlib.pyplot as plt 
import numpy as np
from utils.config import CKPT_PATH_SIMULATION_CONDITIONED_LR5E4_LAST
from utils.config import EXP_FOLDER_TEST, SEED_FOLDER_EXPTOSIM_TEST_32X32
from utils.config import EXP_FOLDER_EXPTOSIM_TEST_FIXED, SEED_FOLDER_EXPTOSIM_TEST_FIXED_32X32
from expcolortoseed_dataset import MyDataset
from models.UNet_Resblocks import UNet_Resblocks
from utils.metrics import point_matching
from matplotlib.font_manager import FontProperties
from utils.config import FPATH

score_font = FontProperties(fname=str(FPATH))

# load the model 
model_path = CKPT_PATH_SIMULATION_CONDITIONED_LR5E4_LAST
model= UNet_Resblocks.load_from_checkpoint(model_path)

# load the test set 
test_dataset= MyDataset(source_folder= EXP_FOLDER_TEST, target_folder= SEED_FOLDER_EXPTOSIM_TEST_32X32,use_vae=True)

# test the performance on the test set
model.eval()
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)


# combine predictions on both these datasets

# files_selected= ['198_1.png', 'Fixed_8_2.TIF','Fixed_6_2.TIF', 'Fixed_7_1.TIF', 'Fixed_12_1.TIF','Fixed_13_2.TIF','Fixed_9_2.TIF', 'Fixed_19_1.TIF']
# combine predictions on both these datasets in specified order

# Select examples; display order is determined by Hungarian tolerance F1
files_selected = ['198_1.png', 'Fixed_8_2.TIF', 'Fixed_6_2.TIF', 'Fixed_7_1.TIF',
                  'Fixed_12_1.TIF', 'Fixed_13_2.TIF', 'Fixed_9_2.TIF', 'Fixed_19_1.TIF']
selected_stems_ordered = [os.path.splitext(f)[0] for f in files_selected]
selected_stems_set     = set(selected_stems_ordered)

all_datasets = [
    MyDataset(source_folder=EXP_FOLDER_TEST,                target_folder=SEED_FOLDER_EXPTOSIM_TEST_32X32,       use_vae=True),
    MyDataset(source_folder=EXP_FOLDER_EXPTOSIM_TEST_FIXED, target_folder=SEED_FOLDER_EXPTOSIM_TEST_FIXED_32X32, use_vae=True),
]

results = {}  # stem -> (source_img, pred_img, target_img, tolerance_f1)
thresh_value = 0.6
matching_tolerance = 4
precision_scores, recall_scores, f1_scores = [], [], []

model.eval()
with torch.no_grad():
    for dataset in all_datasets:
        for i in range(len(dataset)):
            sample = dataset[i]
            source = sample['source']
            target = sample['target']
            stem   = sample['stem']

            source_batch   = source.unsqueeze(0).to(device)
            source_encoded = model.encode_source(source_batch)
            pred           = torch.sigmoid(model(source_encoded))
            pred           = (pred > thresh_value).float().squeeze(0)

            prediction_image = (pred * 255).to(torch.uint8).cpu()
            target_image = (target * 255).round().to(torch.uint8)
            precision, recall, f1 = point_matching(
                prediction_image[0], target_image[0],
                tolerance_radius=matching_tolerance, algorithm='hungarian'
            )
            precision_scores.append(precision)
            recall_scores.append(recall)
            f1_scores.append(f1)

            if stem not in selected_stems_set or stem in results:
                continue

            results[stem] = (
                source_batch[0].permute(1, 2, 0).cpu().numpy(),
                prediction_image.numpy(),
                target_image.numpy(),
                f1,
            )

# Sort the selected examples by unrounded per-image F1, highest to lowest.
ordered_stems = sorted(
    (stem for stem in selected_stems_ordered if stem in results),
    key=lambda stem: results[stem][3], reverse=True,
)
ordered = [results[stem] for stem in ordered_stems]
n = len(ordered)
if n == 0:
    raise ValueError("None of the selected examples were found in the test datasets")

fig, axes = plt.subplots(
    4, n, figsize=(2 * n, 8), squeeze=False,
    gridspec_kw={'height_ratios': [1, 1, 1, 1]},
)
for r, (src_img, pred_img, target_img, f1) in enumerate(ordered):
    axes[0, r].imshow(src_img, cmap='gray' if src_img.shape[2] == 1 else None)
    axes[0, r].axis('off')
    axes[1, r].imshow(pred_img[0], cmap='gray', vmin=0, vmax=255)
    axes[1, r].axis('off')
    axes[2, r].imshow(target_img[0], cmap='gray', vmin=0, vmax=255)
    axes[2, r].axis('off')
    axes[3, r].set_box_aspect(1)
    axes[3, r].text(
        0.5, 0.5, f"{f1:.3f}", ha='center', va='center',
        transform=axes[3, r].transAxes, fontsize=21, fontproperties=score_font,
    )
    axes[3, r].axis('off')
# axes[3, 0].text(
#     -0.08, 0.5, f"Hungarian\ntolerance F1\n({matching_tolerance}px)",
#     ha='right', va='center', transform=axes[3, 0].transAxes, fontsize=11, fontproperties=score_font,
# )

plt.subplots_adjust(wspace=0.02, hspace=0.05)
plt.show()
print(f"Displayed {n} images from highest to lowest Hungarian tolerance F1")
print("Image order:", ", ".join(ordered_stems))
print(
    f"Full test set (n={len(f1_scores)}), threshold={thresh_value}, "
    f"Hungarian tolerance={matching_tolerance}px: "
    f"precision={np.mean(precision_scores):.3f}, "
    f"recall={np.mean(recall_scores):.3f}, "
    f"F1={np.mean(f1_scores):.3f} ± {np.std(f1_scores, ddof=1):.3f}"
)
```
